```python
def laptop_expert(turn_on,internet_working,is_slow):
  if turn_on == "no":
    return "checks the charger and power cable"
  elif internet_working == "no":
   return "checks the wifi router"
  elif is_slow == "yes":
    return "restart the laptop"
  else:
    return "the system doesnot find a common problem"

#input from the user

turn_on = input("Does the laptop turn on? (yes/no): ")
internet_working = input("Is the internet working? (yes/no): ")
is_slow = input("Is the laptop slow? (yes/no):")
advice = laptop_expert(turn_on, internet_working, is_slow)
print("Expert system advice:", advice)
print(advice)




















```
